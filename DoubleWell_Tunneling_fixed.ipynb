# Double‑Well Potential and Quantum Tunnelling
This notebook demonstrates a symmetric double‑well potential and the resulting tunnelling‑split eigenstates. The two lowest bound states form an even (bonding) and odd (anti‑bonding) pair with a small energy splitting.

## 1. Define a symmetric double‑well potential
We use two Gaussian wells centred at (x = ± a). The potential is
V(x) = -V0 * exp[-(x-a)^2/(2σ^2)] - V0 * exp[-(x+a)^2/(2σ^2)]
Parameters (Hartree atomic units):
- depth V0>0 (negative sign makes it a well)
- separation 2a between the wells
- width σ controlling the barrier thickness.

```python
import numpy as np
import matplotlib.pyplot as plt
from functions import make_grid, kinetic_operator, plot_alive

# Physical parameters (Hartree atomic units)
hbar = 1.0
m = 1.0
L = 8.0          # half‑width of the computational box
N_GRID = 2000    # number of grid points

# Build uniform grid
x, dx, _ = make_grid(L, N_GRID)

# Double‑well parameters
V0 = 5.0          # well depth (positive)
a = 2.0           # centre of each well
sigma = 0.5       # width of each Gaussian

def double_well(x):
    return -V0 * np.exp(-((x - a) ** 2) / (2 * sigma ** 2)) \
           - V0 * np.exp(-((x + a) ** 2) / (2 * sigma ** 2))

V = double_well(x)

# Plot the potential (optional)
plt.figure(figsize=(6,3))
plt.plot(x, V, label='Double‑well')
plt.axhline(0, color='k', linewidth=0.5)
plt.xlabel('x')
plt.ylabel('V(x)')
plt.title('Symmetric Double‑Well Potential')
plt.legend()
plt.show()
```

## 2. Build the Hamiltonian and solve
We use the same finite‑difference kinetic operator as in the single‑well examples.

```python
# Kinetic operator (tridiagonal)
T = kinetic_operator(N_GRID, dx, hbar=hbar, m=m)
# Hamiltonian matrix
H = T + np.diag(V)
# Solve for eigenvalues/eigenvectors (few lowest states)
E, psi = np.linalg.eigh(H)
# Ensure ordering
idx = np.argsort(E)
E = E[idx]
psi = psi[:, idx]
print('Lowest eigenvalues (a.u.):', E[:6])
```

## 3. Visualise the lowest eigenstates
The two lowest states should be nearly degenerate, one even and one odd, demonstrating tunnelling splitting.

```python
# Plot the first four eigenfunctions together with the potential
plot_alive(E[:4], psi[:, :4], V, x, no=0)
```

## 4. Analyse tunnelling splitting
The energy difference between the even and odd ground‑state pair is the tunnelling splitting (ΔE = E_odd - E_even).  It decreases exponentially as the barrier height or width increases.

```python
delta_E = E[1] - E[0]
print('Tunnelling splitting ΔE =', delta_E)
```
